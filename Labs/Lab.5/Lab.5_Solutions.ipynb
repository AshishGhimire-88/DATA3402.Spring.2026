1. Create a `matrix` class with the following properties:
    * It can be initialized in 2 ways:
        1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    * Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
    * Matrix assignment works in 2 ways:
        1. If `M_1` and `M_2` are `matrix` instances `M_1=M_2` sets the values of `M_1` to those of `M_2`, if they are the same size. Error otherwise.
        2. In example above `M_2` can be a list of lists of correct size.


```python
class Matrix:
    def __init__(self, data_or_row, col=None):
        if isinstance(data_or_row, int):                                            # Check if the first argument is integer: for knowing if user wants a zero matrix by passing dimension
            row= data_or_row
            if col is None:                                                         # If column is not passed, raise an error since a matrix needs both dimensions
                raise TypeError("Matrix must have both rows and columns")
            self.matrix= [[0]*col for _ in range(row)]                              # Create a 2D list of zeros with 'row' rows and 'col' columns
            self.rows= row
            self.cols=col
        else:                                                                       # Activates if first argumnt is list of list:
            data= data_or_row
            if not all(isinstance(row, list) for row in data):                      # Validation if every element of 'data' is list: for ensuring 'data' is a list of list
                raise TypeError("The Matrix must be List of List")
            
            if not all(len(r)==len(data[0]) for r in data):                         # Validate if every row has same number of element
                raise TypeError("The rows need to have same length")
            
            self.matrix= [row[:] for row in data]                                   # Creating internal matrix independent from the original passed data
            self.rows= len(self.matrix)                                             # Store the number of rows
            self.cols= len(self.matrix[0])                                          # Store the number of columns (length of the first row)

    def __getitem__(self, key):
        if isinstance(key, tuple):                                                  # Validate if 'key' is tuple
            i,j= key
            return self.matrix[i][j]                                                # Return the element at row i and column j of the matrix
        return self.matrix[key]                                                     # If not tuple, return entire row corresponding to 'key'
    
    def __setitem__(self, key, value):
        if isinstance(key, tuple):                                                  # Update Element at row i and column j, if Key is tuple
            i,j= key
            self.matrix[i][j]= value
        else:                                                                       # If the key is not tuple (is single integer) replace row with 'value'
            self.matrix[key]= value
                
    
    def assign(self, other):
        if isinstance(other, Matrix):                                               # Check if 'other' is matrix
            if self.rows!=other.rows or self.cols!=other.cols:                      # Raise error if the shape are not same.
                raise TypeError("Metrices should be of the same size")
            self.matrix=[r[:] for r in other.matrix]                                # Copy the rows of 'other' into 'self'
        
        elif isinstance(other, list):                                               # Check of 'other' is list
            if self.rows!= len(other) or self.cols != len(other[0]):
                raise TypeError("List must be the same size as matrix")             # Raise error if shape of 'self' is not equal to the dimension of 'other'
            self.matrix=[r[:] for r in other]

        else:                                                                       # Raise an error is 'other' is neither a list nor a Matrix
            raise TypeError("Must be List or Matrix!")
    
    def __str__(self):                                                              # Convert each row to string and join it with newlines to print Matrix in readable format.
        return "\n".join(str(row) for row in self.matrix)
```

```python
class Matrix:
    def __init__(self, data_or_row, col=None):
        if isinstance(data_or_row, int):                                            # Check if the first argument is integer: for knowing if user wants a zero matrix by passing dimension
            row= data_or_row
            if col is None:                                                         # If column is not passed, raise an error since a matrix needs both dimensions
                raise TypeError("Matrix must have both rows and columns")
            self.matrix= [[0]*col for _ in range(row)]                              # Create a 2D list of zeros with 'row' rows and 'col' columns
            self.rows= row
            self.cols=col
        else:                                                                       # Activates if first argumnt is list of list:
            data= data_or_row
            if not all(isinstance(row, list) for row in data):                      # Validation if every element of 'data' is list: for ensuring 'data' is a list of list
                raise TypeError("The Matrix must be List of List")
            
            if not all(len(r)==len(data[0]) for r in data):                         # Validate if every row has same number of element
                raise TypeError("The rows need to have same length")
            
            self.matrix= [row[:] for row in data]                                   # Creating internal matrix independent from the original passed data
            self.rows= len(self.matrix)                                             # Store the number of rows
            self.cols= len(self.matrix[0])                                          # Store the number of columns (length of the first row)
    
    def __getitem__(self, key):
        if isinstance(key, tuple):                                                  # When the key is a tuple, e.g. matrix[i, j] or matrix[i:k, j:l] — handles both single element and slice access
            row_part, col_part=key

            if isinstance(row_part, slice):                                         # If the row part is a slice (e.g. 1:3), generate a range of row indices from it
                row_indices= range(*row_part.indices(self.rows))
            else:                                                                   # Otherwise treat it as a single row index, wrapped in a list for uniform processing below
                row_indices= [row_part]
            

            if isinstance(col_part, slice):                                         # If the column part is a slice (e.g. 0:2), generate a range of column indices from it
                col_indices= range(*col_part.indices(self.cols))
            else:                                                                   # Otherwise treat it as a single column index, wrapped in a list for uniform processing below
                col_indices= [col_part]
            

            if not isinstance(row_part, slice) and not isinstance(col_part, slice): # If both row and column are single integers (not slices), return the scalar value directly
                return self.matrix[row_part][col_part]
            
            out= [[self.matrix[i][j] for j in col_indices] for i in row_indices]    # Otherwise, extract the submatrix defined by the row and column ranges, and return it as a Matrix
            return Matrix(out)
        
        else:                                                                       # When the key is not a tuple — handles single-row access or row slicing, e.g. matrix[1] or matrix[1:3]
            if isinstance(key, slice):                                              # If it's a slice, extract all rows in the specified range and return them as a new Matrix
                row_indices= range(*key.indices(self.rows))
                out= [self.matrix[i] for i in row_indices]
                return Matrix(out)
            else:                                                                   # If it's a single integer, return that one row wrapped in a list so it forms a valid 1-row Matrix
                return Matrix([self.matrix[key]])

    def __setitem__(self, key, value):
        if isinstance(key, tuple):                                                  # Update Element at row i and column j, if Key is tuple
            i,j= key
            self.matrix[i][j]= value
        else:                                                                       # If the key is not tuple (is single integer) replace row with 'value'
            self.matrix[key]= value
                
    
    def assign(self, other):
        if isinstance(other, Matrix):                                               # Check if 'other' is matrix
            if self.shape()!= other.shape():                                        # Raise error if the shape are not same.
                raise TypeError("Metrices should be of the same size")
            self.matrix=[r[:] for r in other.matrix]                                # Copy the rows of 'other' into 'self'
        
        elif isinstance(other, list):                                               # Check of 'other' is list
            if self.rows!= len(other) or self.cols != len(other[0]):
                raise TypeError("List must be the same size as matrix")             # Raise error if shape of 'self' is not equal to the dimension of 'other'
            self.matrix=[r[:] for r in other]

        else:                                                                       # Raise an error is 'other' is neither a list nor a Matrix
            raise TypeError("Must be List or Matrix!")



# ======Exercise 2: Implementing Matrix Methods======
    def shape(self):                                                                #Return dimension of matrix as tuple: (row, column)
        return (self.rows, self.cols)
    
    def transpose(self):
        trans = [[0]*self.rows for _ in range(self.cols)]                          # Create a zero matrix of flipped shape: (self.cols, self.rows)
        for i in range(self.rows):
            for j in range(self.cols):
                trans[j][i]=self.matrix[i][j]                                      # Element at (i, j) in 'self' moves to (j, i) in the transposed matrix
        return Matrix(trans)
    
    def row(self, n=0):                                                            # Sets default value to 0: returns first row if 'n' isn't passed
        if n < 0 or n >= self.rows:                                                # Validation of the passed index
            raise ValueError("Out of Index.")
        return Matrix([self.matrix[n]])                                            # Returns the nth row as a Matrix
    
    def column(self, n=0):
        if n < 0 or n >= self.cols:                                                # Validation of the passed index
            raise ValueError("Out of Index")
        out=list()
        for i in range(self.rows):
            out.append([self.matrix[i][n]])                                        # Takes the nth element from each row creating a list of list with a single element in each column
        return Matrix(out)                                                         # Returns the column as a Matrix with shape (rows, 1)
    
    def to_list(self):
        return [row[:] for row in self.matrix]                                     # Returns a list of list version of Matrix using Loop and slice(including every element) method
    
    def block(self, m_0, m_1,  n_0, n_1):
        if n_0 <0 or n_1>self.cols or m_0<0 or m_1>self.rows:                      # Validate if rows(m_0, m_1) and columns(n_0, n_1) are within the bound of Matric
            raise ValueError("Incides for Block's Out of Range.")               
        out= list()
        for i in range(m_0, m_1):                                                  # Iterate over selected row: m_0 to m_1(excluded)
            row=[]
            for j in range(n_0, n_1):                                              # Iterate over selected columns: n_0 to n_1(excluded)
                row.append(self.matrix[i][j])                                      # Create a list of elements on row i and colmn j
            out.append(row)                                                        # Append the completed row to the output list, building up the submatrix
        return Matrix(out)
    
    def __str__(self):                                                             # Convert each row to a string and join them with newlines so the matrix prints in a readable grid format
        return "\n".join(str(row) for row in self.matrix)
```

3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    * `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    * `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    * `eye(n)`: returns the n by n identity matrix.

```python
def constant(n,m,c):        #Generate constant matrix of dimension n*m
    return Matrix([[c]*m for _ in range(n)])

def zeros(n,m):             #Generate zero matrix of dimension n*m
    return Matrix([[0]*m for _ in range(n)])

def ones(n,m):             #Generate ones matrix of dimension n*m
    return Matrix([[1]*m for _ in range(n)])

def eye(n):                 #Geenrate identity matrix of dimension n*n
    x = zeros(n,n)
    for i in range(n):
        x.matrix[i][i]=1
    return x
```

```python
class Matrix:
    def __init__(self, data_or_row, col=None):
        if isinstance(data_or_row, int):                                            # Check if the first argument is integer: for knowing if user wants a zero matrix by passing dimension
            row= data_or_row
            if col is None:                                                         # If column is not passed, raise an error since a matrix needs both dimensions
                raise TypeError("Matrix must have both rows and columns")
            self.matrix= [[0]*col for _ in range(row)]                              # Create a 2D list of zeros with 'row' rows and 'col' columns
            self.rows= row
            self.cols=col
        else:                                                                       # Activates if first argumnt is list of list:
            data= data_or_row
            if not all(isinstance(row, list) for row in data):                      # Validation if every element of 'data' is list: for ensuring 'data' is a list of list
                raise TypeError("The Matrix must be List of List")
            
            if not all(len(r)==len(data[0]) for r in data):                         # Validate if every row has same number of element
                raise TypeError("The rows need to have same length")
            
            self.matrix= [row[:] for row in data]                                   # Creating internal matrix independent from the original passed data
            self.rows= len(self.matrix)                                             # Store the number of rows
            self.cols= len(self.matrix[0])                                          # Store the number of columns (length of the first row)
    
    def __getitem__(self, key):
        if isinstance(key, tuple):                                                  # When the key is a tuple, e.g. matrix[i, j] or matrix[i:k, j:l] — handles both single element and slice access
            row_part, col_part=key

            if isinstance(row_part, slice):                                         # If the row part is a slice (e.g. 1:3), generate a range of row indices from it
                row_indices= range(*row_part.indices(self.rows))
            else:                                                                   # Otherwise treat it as a single row index, wrapped in a list for uniform processing below
                row_indices= [row_part]
            

            if isinstance(col_part, slice):                                         # If the column part is a slice (e.g. 0:2), generate a range of column indices from it
                col_indices= range(*col_part.indices(self.cols))
            else:                                                                   # Otherwise treat it as a single column index, wrapped in a list for uniform processing below
                col_indices= [col_part]
            

            if not isinstance(row_part, slice) and not isinstance(col_part, slice): # If both row and column are single integers (not slices), return the scalar value directly
                return self.matrix[row_part][col_part]
            
            out= [[self.matrix[i][j] for j in col_indices] for i in row_indices]    # Otherwise, extract the submatrix defined by the row and column ranges, and return it as a Matrix
            return Matrix(out)
        
        else:                                                                       # When the key is not a tuple — handles single-row access or row slicing, e.g. matrix[1] or matrix[1:3]
            if isinstance(key, slice):                                              # If it's a slice, extract all rows in the specified range and return them as a new Matrix
                row_indices= range(*key.indices(self.rows))
                out= [self.matrix[i] for i in row_indices]
                return Matrix(out)
            else:                                                                   # If it's a single integer, return that one row wrapped in a list so it forms a valid 1-row Matrix
                return Matrix([self.matrix[key]])

    def __setitem__(self, key, value):
        if isinstance(key, tuple):                                                  # Update Element at row i and column j, if Key is tuple
            i,j= key
            self.matrix[i][j]= value
        else:                                                                       # If the key is not tuple (is single integer) replace row with 'value'
            self.matrix[key]= value
                
    
    def assign(self, other):
        if isinstance(other, Matrix):                                               # Check if 'other' is matrix
            if self.shape()!= other.shape():                                        # Raise error if the shape are not same.
                raise TypeError("Metrices should be of the same size")
            self.matrix=[r[:] for r in other.matrix]                                # Copy the rows of 'other' into 'self'
        
        elif isinstance(other, list):                                               # Check of 'other' is list
            if self.rows!= len(other) or self.cols != len(other[0]):
                raise TypeError("List must be the same size as matrix")             # Raise error if shape of 'self' is not equal to the dimension of 'other'
            self.matrix=[r[:] for r in other]

        else:                                                                       # Raise an error is 'other' is neither a list nor a Matrix
            raise TypeError("Must be List or Matrix!")
    
    def shape(self):                                                                #Return dimension of matrix as tuple: (row, column)
        return (self.rows, self.cols)
    
    def transpose(self):
        trans = [[0]*self.rows for _ in range(self.cols)]                          # Create a zero matrix of flipped shape: (self.cols, self.rows)
        for i in range(self.rows):
            for j in range(self.cols):
                trans[j][i]=self.matrix[i][j]                                      # Element at (i, j) in 'self' moves to (j, i) in the transposed matrix
        return Matrix(trans)
    
    def row(self, n=0):                                                            # Sets default value to 0: returns first row if 'n' isn't passed
        if n < 0 or n >= self.rows:                                                # Validation of the passed index
            raise ValueError("Out of Index.")
        return Matrix([self.matrix[n]])                                            # Returns the nth row as a Matrix
    
    def column(self, n=0):
        if n < 0 or n >= self.cols:                                                # Validation of the passed index
            raise ValueError("Out of Index")
        out=list()
        for i in range(self.rows):
            out.append([self.matrix[i][n]])                                        # Takes the nth element from each row creating a list of list with a single element in each column
        return Matrix(out)                                                         # Returns the column as a Matrix with shape (rows, 1)
    
    def to_list(self):
        return [row[:] for row in self.matrix]                                     # Returns a list of list version of Matrix using Loop and slice(including every element) method
    
    def block(self, m_0, m_1,  n_0, n_1):
        if n_0 <0 or n_1>self.cols or m_0<0 or m_1>self.rows:                      # Validate if rows(m_0, m_1) and columns(n_0, n_1) are within the bound of Matric
            raise ValueError("Incides for Block's Out of Range.")               
        out= list()
        for i in range(m_0, m_1):                                                  # Iterate over selected row: m_0 to m_1(excluded)
            row=[]
            for j in range(n_0, n_1):                                              # Iterate over selected columns: n_0 to n_1(excluded)
                row.append(self.matrix[i][j])                                      # Create a list of elements on row i and colmn j
            out.append(row)                                                        # Append the completed row to the output list, building up the submatrix
        return Matrix(out)
    
    def scalarmul(self, c):
        out= zeros(self.rows, self.cols)                                           # Create a zero matrix of the same size to store the result
        for i in range(self.rows):
            for j in range(self.cols):
                out.matrix[i][j]= c*self.matrix[i][j]                              # Multiply each element of the matrix by the scalar c and store the result at the same position in the output matrix
        return out
    
    def add(self, other):
        if self.shape() != other.shape():                                          # Check that both matrices have same shape: Addition is only possible if matrices have same shape
            raise TypeError("Matrices must be of same dimension")
        out= zeros(self.rows, self.cols)                                           # Create a zero matrix of the same size
        for i in range(self.rows):
            for j in range(self.cols):
                out.matrix[i][j]= self.matrix[i][j] + other.matrix[i][j]           # Add corresponding elements from both matrices and store the result in the 'out' Matrix
        return out
    
    def sub(self, other):
        if self.shape() != other.shape():                                          # Check that both matrices have same shape: Substraction is only possible if matrices have same shape
            raise TypeError("Matrices must be of same dimension")
        out= zeros(self.rows, self.cols)
        for i in range(self.rows):
            for j in range(self.cols):
                out.matrix[i][j]= self.matrix[i][j] - other.matrix[i][j]
        return out
    
    def mat_mult(self, other):
        if self.cols != other.rows:                                                # Validate if number of columns in first Matrix is equal to number of rows in second Matrix
            raise TypeError("Matrix Multiplication not possible!")
        out= zeros(self.rows, other.cols)                                          # Create a zero matrix of shape (self.rows, other.cols): Resulting shape after Matrix multiplication 
        for i in range(self.rows):
            for j in range(other.cols):
                for k in range(self.cols):
                    out.matrix[i][j]+=self.matrix[i][k]*other.matrix[k][j]         # Implementation of Dot Product
        return out
    
    def elem_mult(self, other):
        if self.shape() != other.shape():                                          # Check that both matrices have same shape: Element-wise multiplication requires matrices to have same shape
            raise TypeError("Matrices must be of same dimension")
        out= zeros(self.rows, self.cols)                                           # Create a zero matrix of the same size to store the result
        for i in range(self.rows):
            for j in range(self.cols):
                out.matrix[i][j]= self.matrix[i][j] * other.matrix[i][j]           # Multiply corresponding elements from both matrices and store the result in the output matrix
        return out
    
    def equals(self, other):
        if self.shape() != other.shape():                                          # Validate if matrices have same shape: Equal matrices should have same shape
            return False
        for i in range(self.rows):                                                 # Loop through every row and column to compare each corresponding pair of elements
            for j in range(self.cols):
                if self.matrix[i][j]!=other.matrix[i][j]:
                    return False                                                   # If at least 1 element is not same, return False
        return True                                                                # Retrun True if all elements are same
    

# ======Excercise 5: Overloading of Operators======

    # Multiply by Constant: 
    def __mul__(self, other):
        if isinstance(other, (int, float)):                                        # Handels Matrix * Cosntant
            return self.scalarmul(other)
        elif isinstance(other, Matrix):                                            # Handels Matrix Multiplication   
            return self.mat_mult(other)
        else:
            raise TypeError("Multiplication not supported!")
    
    def __rmul__(self, other):                                                     # Handels Constant * Matrix
        return self.scalarmul(other)

    def __add__(self, other):                                                      # Adds Matrices
        return self.add(other)
    

    def __sub__(self, other):                                                      # Substract Matrices
        return self.sub(other)
    

    def __matmul__(self, other):                                                   # Multiply Matrices
        return self.mat_mult(other)
    

    def __eq__(self, other):                                                       # Check if Matrices are equal
        return self.equals(other)
    
    
    def __str__(self):                                                             # Convert each row to a string and join them with newlines so the matrix prints in a readable grid format
        return "\n".join(str(row) for row in self.matrix)
```
